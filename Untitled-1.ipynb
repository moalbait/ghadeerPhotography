```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Data
data = {
    'Weight1': [50, 60, 70, 80, 90, 100, 110, 120, 130, 140],
    'Height1': [225, 160, 200, 11, 50, 150, 210, 100, 120, 230],
    'Weight2': [55, 65, 75, 85, 95, 105, 115, 125, 135, 145],
    'Height2': [50, 165, 70, 185, 90, 165, 215, 111, 101, 150],
    'Weight3': [52, 62, 72, 82, 92, 102, 112, 122, 132, 142],
    'Height3': [152, 7, 45, 182, 192, 202, 50, 148, 80, 135]
}

df = pd.DataFrame(data)

x = np.arange(len(df))  # x positions
width = 0.2  # bar width

# Grouped bars
plt.bar(x - width*1.5, df['Weight1'], width, color='r', label='Weight1')
plt.bar(x - width*0.5, df['Height1'], width, color='pink', label='Height1')

plt.bar(x + width*0.5, df['Weight2'], width, color='g', label='Weight2')
plt.bar(x + width*1.5, df['Height2'], width, color='lightgreen', label='Height2')

# Optional: if you want Weight3 and Height3 too
# plt.bar(x + width*2.5, df['Weight3'], width, color='b', label='Weight3')
# plt.bar(x + width*3.5, df['Height3'], width, color='lightblue', label='Height3')

plt.title('Weight and Height Distribution')
plt.xlabel('Index')
plt.ylabel('Value')
plt.xticks(x, [f"#{i+1}" for i in x])
plt.grid(axis='y')
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Create DataFrame with 3 bars
data = {
    'Group': ['Group 1', 'Group 2', 'Group 3'],
    'Value': [120, 180, 90]  # values for each bar
}

df= pd.DataFrame(data)
# Create bar plot
plt.bar(df['Group'], df['Value'],color=['r', 'g', 'b'])
plt.title('Group Value Bar Plot')
plt.xlabel('Group')
plt.ylabel('Value')

#show value grid lines
for i, v in enumerate(df['Value']):
    plt.text(i, v + 3, str(v), ha='center', va='bottom')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42)
groups = ['Group 1' , 'Group 2' , 'Group 3 ']
df = pd.DataFrame({
                    # generate 10 group labels for each group
                    'Group': np.repeat(groups, 10),

                    # generate 10 weights for each group
                    'weight': np.concat([np.random.randint(55, 85, 10),#group 1
                                         np.random.randint(60,91, 10), #group 2
                                         np.random.randint(65,88,10)]), #group 3

                    # generate 10 heights for each group
                    'height': np.concat([ np.random.randint(100,180,10),#group 1
                                          np.random.randint(100,1////0,10),#group 2
                                          np.random.randint(130,170,10)]) #group 3
})
group_A= df[df['Group']=='Group 1']
group_B= df[df['Group']=='Group 2']
group_C= df[df['Group']=='Group 3 ']

plt.scatter(group_A['weight'], group_A['height'], color='r', label='Group 1')
plt.scatter(group_B['weight'],group_B['height'],color='g',label= 'Group 2')
plt.scatter(group_C['height'],group_C['height'],color='y',label= 'Group 3')

plt.title('Weights vs Heights')
plt.xlabel('Weight')
plt.ylabel('Height')
plt.grid(axis='y',linestyle='--',alpha= 0.5)
```

```python
import sqlite3

# Connect to a database file (creates it if it doesn't exist)
conn = sqlite3.connect("mydb.db")
cur = conn.cursor()

# Drop table if exists (fresh start)
cur.execute("DROP TABLE IF EXISTS users")

# Create table
cur.execute("""
CREATE TABLE users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT,
    age INTEGER
)
""")

# Insert rows
cur.executemany("""
INSERT INTO users (name, age) VALUES (?, ?)
""", [
    ('Alice', 30),
    ('Bob', 25),
    ('Mike', 32)
])
cur.execute("INSERT INTO users (name, age) VALUES (?, ?)", ('David', 28))
cur.execute("UPDATE users SET age = ? WHERE name = ?", (31, 'Alice'))
conn.commit()

# Commit changes
conn.commit()

# Query the table
cur.execute("SELECT * FROM users")
rows = cur.fetchall()

# Display results
for row in rows:
    print(row)

# Close connection
conn.close()
```

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

# -------------------------------
# 1️⃣ Connect to the database
# -------------------------------
conn = sqlite3.connect("mydb.db")

# -------------------------------
# 2️⃣ Load data into pandas DataFrame
# -------------------------------
df = pd.read_sql_query("SELECT * FROM users", conn)

# -------------------------------
# 3️⃣ Bar chart: age of each user
# -------------------------------
plt.figure(figsize=(8,5))
plt.bar(df['name'], df['age'], color='skyblue')
plt.title("Age of Each User")
plt.xlabel("Name")
plt.ylabel("Age")
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Show values on top of bars
for i, age in enumerate(df['age']):
    plt.text(i, age + 0.5, str(age), ha='center', fontweight='bold')

plt.show()

# -------------------------------
# 4️⃣ Optional: Histogram of ages
# -------------------------------
plt.figure(figsize=(6,4))
plt.hist(df['age'], bins=5, color='lightgreen', edgecolor='black')
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()

# -------------------------------
# 5️⃣ Close the connection
# -------------------------------
conn.close()
```
